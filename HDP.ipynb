## 11. Train-Test Split
Train-test splitting is an important step in
machine learning. It divides the dataset into
training and testing sets.

The training set is used to train the machine
learning model, while the testing set is used
to evaluate its performance on unseen data.

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('condition', axis=1)
y = df['condition']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

#### Summary
In this step, we divided the dataset into training and testing sets using the train_test_split() function from Scikit-learn. We used a 70:30 ratio, where 70% of the data (207 records) was allocated for training and 30% (90 records) for testing. The stratify parameter maintained the class distribution in both sets. This split prepares the data for training and evaluating machine learning models.

## 10.1. Feature Selection

```python
from sklearn.feature_selection import SelectKBest, f_classif
```

```python
selector = SelectKBest(score_func=f_classif, k=10)

X_train_selected = selector.fit_transform(X_train, y_train)
X_test_selected = selector.transform(X_test)
```

```python
selected_features = X.columns[selector.get_support()]

print("Selected Features:")
print(selected_features.tolist())

print("\nTotal Selected Features:", len(selected_features))
```

```python
feature_scores = pd.DataFrame({
    'Feature': X.columns,
    'Score': selector.scores_
})

feature_scores = feature_scores.sort_values(
    by='Score', ascending=False
)

print(feature_scores)
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=feature_scores,
    x='Score',
    y='Feature'
)

plt.title('Feature Selection Scores')
plt.xlabel('ANOVA F-Score')
plt.ylabel('Features')

plt.show()
```
